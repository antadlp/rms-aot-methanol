# Which molecules were used?

The molecules of the systems studied were chosen thinking in reverse micelles (RMs) coexisting with alcohols. We selected AOT/isooctane/water for the RMs, which is the most common combination, and methanol for the alcohol.

[1. AOT](#aot_ff) <br>
[2. Na+](#na_ff) <br>
[3. Isooctane](#isooctane_ff) <br>
[4. Methanol](#methanol_ff) <br>
[5. Water](#water_ff)

# 1.  AOT

Aerosol OT (sodium diethylhexyl sulfosuccinate, sodium docusate, AOT) is the most used surfactant for its efficiency.[[1]](https://doi.org/10.1016/0001-8686(95)80005-N)[[2]](https://doi.org/10.1021/cr200254q)[[3]](https://doi.org/10.1021/la000342i). It has the following structure

## 1.1 AOT structure (nglview)

```python
import nglview as nv

PATH_AOT = "../data/pdbs/63UD_allatom_optimised_geometry.pdb"

view = nv.show_file(PATH_AOT)
view.center()
view
```

```python
view.render_image()
view._display_image()
```

<a id='aot_ff'></a>
## 1.2 AOT forcefield

The AOT forcefield and its optimized geometry were obtained from [Automated Topology Builder (ATB) and Repository Version 3.0](https://atb.uq.edu.au/)

The Molid is: 25804

The topology .itp and geometry .pdb are included in this repository

[aot topology](../data/itps/63UD_GROMACS_G54A7FF_allatom.itp) <br>
[aot geometry](../data/pdbs/63UD_allatom_optimised_geometry.pdb)

# 2.  Na+ 

Na+ is the counterion of the hydrophilic headgroup in the RMs formed by Aerosol-OT (AOT) [[1]](https://doi.org/10.1021/acs.jpclett.8b03389)

The Na+ forcefield and its optimized geometry were obtained from [Automated Topology Builder (ATB) and Repository Version 3.0](https://atb.uq.edu.au/)

The Molid is: 307652

The topology .itp and geometry .pdb are included in this repository

[Na+ topology](../data/itps/E0XM_GROMACS_G54A7FF_allatom.itp) <br>
[Na+ geometry](../data/pdbs/E0XM_allatom_optimised_geometry.pdb)

# 3. Isooctane

Because this is the solvent media, we decided to use a united-atom approximation to model this molecule.

## 3.1 Isooctane structure (nglview)

```python
import nglview as nv

PATH_ISOO = "../data/pdbs/G016_unitedatom_optimised_geometry.pdb"

view = nv.show_file(PATH_ISOO)
view.center()
view
```
